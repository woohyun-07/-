```python

from collections import deque

# 그래프 구성
graph = {
    '안우현': ['최우진', '배정우', '이가을','권태현'],
    '최우진': [],
    '배정우': [],

    '이가을': ['김준석', '강석민','윤여주'],
    '김준석': [],
    '강석민': [],

    '권태현': ['김주홍', '염도형'],
    '김주홍': [],
    '염도형': [],

    '박채완': ['이도현', '이대환'],
    '이도현': [],
    '이대환': [],

    '정민석': ['이정수', '김진성'],
    '이정수': [],
    '김진성': [],

    '윤여주': ['김수인', '황정민'],
    '김수인': [],
    '황정민': []
}


# 1. DFS (깊이 우선 탐색)
def dfs(graph, v, visited):
    visited.add(v)
    print(v, end=' ')

    for neighbor in graph[v]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


print("DFS 탐색 결과:")
dfs(graph, '안우현', set())


# 2. BFS (너비 우선 탐색)
def bfs(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    while queue:
        v = queue.popleft()
        print(v, end=' ')

        for neighbor in graph[v]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


print("\n\nBFS 탐색 결과:")
bfs(graph, '안우현')


# 3. 친구 추천 시스템
def recommend_friend(graph, start):
    visited = set()
    queue = deque([(start, 0)])

    visited.add(start)

    print(f"\n\n{start}의 친구 추천:")

    while queue:
        current, depth = queue.popleft()

        if depth == 2:
            print(current)

        if depth < 2:
            for neighbor in graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append((neighbor, depth + 1))


recommend_friend(graph, '안우현')
```
